```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

# Thay thế bằng API Key của bạn
SCRAPERAPI_KEY = "77d67b0df073c08c22c4d64ad8ebc1f3"

# URL cần lấy dữ liệu
url = "https://www.digikey.com/en/products/result?s=N4IgTCBcDaIO4EsAWBDEBdAvkA"
org_url = "https://www.digikey.com"

# Dùng ScraperAPI để gửi request
payload = {
    "api_key": SCRAPERAPI_KEY,
    "url": url
}
response = requests.get("http://api.scraperapi.com/", params=payload)

# Kiểm tra nếu request thành công
if response.status_code != 200:
    print("❌ Không lấy được dữ liệu! Kiểm tra API Key hoặc IP có bị chặn không.")
    exit()

# Parse HTML bằng BeautifulSoup
soup = BeautifulSoup(response.text, "html.parser")

# Lấy danh sách sản phẩm
list_type_product = soup.find_all("div", attrs={"data-testid": "card"})

# Kiểm tra nếu không có sản phẩm
if not list_type_product:
    print("❌ Không tìm thấy sản phẩm!")
    exit()

# Trích xuất dữ liệu sản phẩm
data_digikey = []
for product in list_type_product:
    link_tag = product.find('a', attrs={'rel': 'opener'})
    link = org_url + link_tag["href"] if link_tag else "Không có link"

    image_tag = product.find('img')
    link_image = image_tag['src'] if image_tag else "Không có ảnh"

    product_name=product.find("div",attrs={'class':'tss-css-1nopv3b-categoryLabel'})
    product_name=product_name.text.strip()

    product_texts = product.find_all("div", attrs={'class': 'tss-css-cuw97q-categoryText'})

    product_type = product_texts[0].text.strip() if len(product_texts) > 0 else "Không có loại"
    product_count_item = product_texts[1].text.strip() if len(product_texts) > 1 else "Không có số lượng"
    data_digikey.append([product_name,product_type,product_count_item,link, link_image])

# Chuyển dữ liệu sang DataFrame
df_digikey = pd.DataFrame(data_digikey, columns=["Product Name","Product Type","Count Item","Product Link", "Image Link"])
print(df_digikey)
df_digikey.to_csv("digikey.csv", index=False, encoding="utf-8")
print("Lưu file csv thành công")
```
